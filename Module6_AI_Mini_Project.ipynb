```python
quotes = {
    "happy": "Keep smiling! Your positivity inspires everyone.",
    "sad": "Every difficult day brings you closer to a better one.",
    "tired": "Rest if you must, but never quit.",
    "success": "Success comes to those who keep learning.",
    "default": "Believe in yourself. You can achieve anything!"
}

mood = input("Enter your mood: ").lower()

print("\nMotivational Quote:")
print(quotes.get(mood, quotes["default"]))
```
